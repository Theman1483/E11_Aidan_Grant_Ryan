```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#setting up dataframes
indata = pd.read_csv('https://raw.githubusercontent.com/Theman1483/E11_Aidan_Grant_Ryan/refs/heads/main/indoortest.csv')
outdata = pd.read_csv('https://raw.githubusercontent.com/Theman1483/E11_Aidan_Grant_Ryan/refs/heads/main/outdoortest.csv')
mistdata = pd.read_csv('https://raw.githubusercontent.com/Theman1483/E11_Aidan_Grant_Ryan/refs/heads/main/newoutdoortest.csv')
```

```python
indata25, indatatemp, indatahum, indatap = indata.loc[1:, 'particles 2.5um'], indata.loc[1:, 'Temperature'], indata.loc[1:, 'Humidity'], indata.loc[1:, 'Pressure']
outdata25, outdatatemp, outdatahum, outdatap = outdata.loc[1:, 'particles 2.5um'], outdata.loc[1:, 'Temperature'], outdata.loc[1:, 'Humidity'], outdata.loc[1:, 'Pressure']
mistdata25, mistdatatemp, mistdatahum, mistdatap = mistdata.loc[90:, 'particles 2.5um'], mistdata.loc[90:, 'Temperature'], mistdata.loc[90:, 'Humidity'], mistdata.loc[90:, 'Pressure']
```

- Hummidity and temperature seem to be vaguely related in some enviroments, but not absolutely as seen in the graph of mist temperature vs humidity and at the ends of the other data as well.
- 2.5um particles and temperature don't seem related at all.

```python
%matplotlib inline

plt.rcParams["figure.figsize"] = [12,30]

plt.subplot(5,2,1)
plt.plot(indatatemp, indata25)
plt.title("Temperature vs 2.5um Particles (Inside)")
plt.xlabel('Temperature (c)')
plt.ylabel('2.5um Particles')

plt.subplot(5,2,2)
plt.plot(indatatemp, indatahum)
plt.title("Temperature vs Hummidity (Inside)")
plt.xlabel('Temperature (c)')
plt.ylabel('Humidity')

plt.subplot(5,2,3)
plt.plot(outdatatemp, outdata25)
plt.title("Temperature vs 2.5um Particles (Outside)")
plt.xlabel('Temperature (c)')
plt.ylabel('2.5um Particles')

plt.subplot(5,2,4)
plt.plot(outdatatemp, outdatahum)
plt.title("Temperature vs Hummidity (Outside)")
plt.xlabel('Temperature (c)')
plt.ylabel('Humidity')

plt.subplot(5,2,5)
plt.plot(mistdatatemp, mistdata25)
plt.title("Temperature vs 2.5um Particles (Vent)")
plt.xlabel('Temperature (c)')
plt.ylabel('2.5um Particles')

plt.subplot(5,2,6)
plt.scatter(mistdatatemp, mistdatahum, s=10)
plt.title("Temperature vs Hummidity (Vent)")
plt.xlabel('Temperature (c)')
plt.ylabel('Humidity')

plt.subplot(5,2,7)
plt.plot(indatatemp, indatap)
plt.title("Temperature vs Pressure (Inside)")
plt.xlabel('Temperature (c)')
plt.ylabel('Pressure')

plt.subplot(5,2,8)
plt.scatter(mistdatatemp, mistdatap, s=1)
plt.title("Temperature vs Pressure (Vent)")
plt.xlabel('Temperature (c)')
plt.ylabel('Pressure')

```
